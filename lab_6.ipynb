```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.metrics import mean_squared_error

df = pd.read_csv('datalab6.csv')

x = df[['x']]
y = df['y']

x_train = x.iloc[:5]
x_valid = x.iloc[5:7]
x_test = x.iloc[7:]

y_train = y.iloc[:5]
y_valid = y.iloc[5:7]
y_test = y.iloc[7:]

# Linear Regression
lr_model = LinearRegression()
lr_model.fit(x_train, y_train)

# Lasso Regression (using default alpha=1.0 for demonstration)
lasso_model = Lasso(alpha=1.0)
lasso_model.fit(x_train, y_train)

# Ridge Regression (using default alpha=1.0 for demonstration)
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(x_train, y_train)

# Print coefficients and intercepts
print("--- Linear Regression ---")
print("Slope M:", lr_model.coef_[0])
print("Intercept C:", lr_model.intercept_)

print("\n--- Lasso Regression ---")
print("Slope M:", lasso_model.coef_[0])
print("Intercept C:", lasso_model.intercept_)

print("\n--- Ridge Regression ---")
print("Slope M:", ridge_model.coef_[0])
print("Intercept C:", ridge_model.intercept_)

# Evaluate Models
models = {
    'Linear Regression': lr_model,
    'Lasso': lasso_model,
    'Ridge': ridge_model
}

for name, model in models.items():
    print(f"\n=== {name} Performance ===")
    y_train_pred = model.predict(x_train)
    print("Training MSE:", mean_squared_error(y_train, y_train_pred))

    y_valid_pred = model.predict(x_valid)
    print("Validating MSE:", mean_squared_error(y_valid, y_valid_pred))

    y_test_pred = model.predict(x_test)
    print("Testing MSE:", mean_squared_error(y_test, y_test_pred))

# Plotting
plt.figure(figsize=(10, 6))
plt.scatter(x_train, y_train, color='blue', label='Training Data', zorder=5)
plt.scatter(x_valid, y_valid, color='orange', label='Validating Data', zorder=5)
plt.scatter(x_test, y_test, color='red', label='Testing Data', zorder=5)

plt.plot(x, lr_model.predict(x), color='green', linewidth=2, label='Linear Regression')
plt.plot(x, lasso_model.predict(x), color='purple', linewidth=2, linestyle='--', label='Lasso Regression')
plt.plot(x, ridge_model.predict(x), color='cyan', linewidth=2, linestyle=':', label='Ridge Regression')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Regularization Comparison')
plt.legend()
plt.show()
```
