```python
import pandas as pd
df=pd.read_csv("mydata.csv")
print(df)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Clean column names by stripping whitespace
df.columns = df.columns.str.strip()

# Print all columns to ensure 'Department' is correctly identified
print("DataFrame Columns after stripping whitespace:", df.columns.tolist())

# Calculate the value counts for the 'Department' column
department_counts = df['Department'].value_counts()

# Get the labels (department names) and sizes (counts) for the pie chart
labels = department_counts.index
sizes = department_counts.values

# Create the pie chart
plt.figure(figsize=(8, 8)) # Optional: set figure size for better visualization
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Distribution of Employees by Department')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```
